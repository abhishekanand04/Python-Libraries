#concatenation

```python
import pandas as pd
import numpy as np
```

```python
users = pd.DataFrame({'userid':[1,2,3],'name':['nikhil','puneeth','dhiraj']})
users
```

```python
msgs = pd.DataFrame({'userid':[1,1,2,4],'msg':['hmm','acha','theek hai','nice']})
msgs
```

```python
pd.concat([users,msgs]) #default axis=0
```

```python
df3 = pd.concat([users,msgs],axis=1)
```
